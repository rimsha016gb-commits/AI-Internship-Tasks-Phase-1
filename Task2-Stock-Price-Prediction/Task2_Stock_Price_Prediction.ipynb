# Task 2: Predict Future Stock Prices (Short-Term)

## Problem Statement

The purpose of this task is to predict the next day's closing stock price using historical market data and a Linear Regression model.

## Goal

- Load historical stock data using the yfinance library.
- Inspect and understand the dataset structure.
- Prepare features and target for prediction.
- Train a Linear Regression model.
- Evaluate the model using MAE and RMSE.
- Plot actual vs predicted closing prices.

## Importing Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

## Loading the Dataset

Apple (AAPL) stock data is loaded using the yfinance library covering 2 years (2022-2024).

```python
# Download Apple stock data from Yahoo Finance
df = yf.download('AAPL', start='2022-01-01', end='2024-01-01')

df.head()
```

## Dataset Inspection

Checking dataset shape, columns and basic information.

```python
print("Shape:", df.shape)
print(df.columns)
df.info()
```

## Data Preprocessing

Selecting relevant features, creating the target column (next day's close price), and checking for missing values.

```python
# Select relevant columns
df = df[['Open', 'High', 'Low', 'Volume', 'Close']].copy()

# Check for missing values
print(df.isnull().sum())

# Create target: next day's closing price
df['Target'] = df['Close'].shift(-1)

# Drop last row which has no target
df.dropna(inplace=True)

print("\nShape after preprocessing:", df.shape)
df.head()
```

## Statistical Summary

```python
df.describe()
```

## Train / Test Split

Splitting data into 80% training and 20% testing. shuffle=False preserves the time order.

```python
X = df[['Open', 'High', 'Low', 'Volume']]
y = df['Target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

print("Training samples:", X_train.shape[0])
print("Testing samples: ", X_test.shape[0])
```

## Model Training

Training a Linear Regression model on the training data.

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Model trained successfully.")
```

## Model Evaluation

Evaluating predictions using Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE).

```python
y_pred = model.predict(X_test)

mae  = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
```

## Actual vs Predicted Plot

Visualizing how closely the predicted prices follow the actual closing prices.

```python
plt.figure(figsize=(12, 5))
plt.plot(y_test.values, label='Actual Close Price', color='blue')
plt.plot(y_pred,        label='Predicted Close Price', color='orange', linestyle='--')
plt.title('AAPL - Actual vs Predicted Next-Day Closing Price')
plt.xlabel('Test Sample Index')
plt.ylabel('Price (USD)')
plt.legend()
plt.tight_layout()
plt.show()
```

# Final Insights

### Observations

1. The Linear Regression model performs well for short-term stock price prediction.
2. Open, High, and Low prices are the strongest predictors of the next day's closing price.
3. Volume alone has less predictive power compared to price-based features.
4. The predicted price line closely tracks the actual price line in the plot.
5. A few larger deviations occur during periods of high market volatility.

### Conclusion

The model achieves a low MAE (~1.87) and RMSE (~2.39) on Apple stock data, which represents roughly 1% error on a stock priced around $150-$180. This confirms that price-based features (Open, High, Low) carry strong information about the next day's closing price and Linear Regression is a solid baseline model for short-term stock prediction.
